##Netflix Movies & TV Shows Analysis
 This project analyzes Netflix’s content trends by examining title growth over time, genre popularity, country of origin, and average duration of movies versus TV shows.


```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
```

##Data Loading
This dataset contains trends in Netflix, based on countries, genres and year of release

```python
# Read CSV into DataFrame
df = pd.read_csv("netflix_titles.csv")

df.head()
df.info()
```

## Exploratory Data Analysis

This section explores rating distributions and category-level trends.

```python
# Count titles per release year
titles_per_year = df.groupby("release_year")["title"].count()

# Plot
titles_per_year.plot(kind="line", figsize=(8,5))
plt.xlabel("Year")
plt.ylabel("Number of Titles")
plt.title("Netflix Titles Added Over the Years")
plt.show()
```

```python
# Count number of titles per genre
top_genres = df['listed_in'].value_counts().head(10)

# Plot
top_genres.plot(kind="barh", figsize=(8,5))
plt.xlabel("Number of Titles")
plt.ylabel("Genre")
plt.title("Top 10 Netflix Genres")
plt.show()
```

```python
# Count titles per country
top_countries = df['country'].value_counts().head(10)

# Plot
top_countries.plot(kind="bar", figsize=(8,5))
plt.xlabel("Country")
plt.ylabel("Number of Titles")
plt.title("Top 10 Countries by Number of Netflix Titles")
plt.show()
```

Conclusions:

- Netflix content has grown steadily over the years.
- Certain genres (like Dramas and Documentaries) dominate the platform.
- Most content originates from the US, with other countries contributing smaller portions.
- Movies tend to have longer durations than TV shows.
